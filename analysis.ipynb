```python
#Cleaning the Reviews column
df['Reviews'] = df['Reviews'].astype(str)


df['Reviews'] = df['Reviews'].str.replace(',', '', regex=False)


def reviews_to_int(x):
    try:
        if 'M' in x:
            return int(float(x.replace('M','')) * 1_000_000)
        else:
            return int(float(x))  
    except:
        return np.nan

df['Reviews'] = df['Reviews'].apply(reviews_to_int)


df['Reviews'].head(10841)
```

```python
#Inspect Reviews
df['Reviews'].describe()
```

```python
#Cleaning the Installs column
df['Installs'] = df['Installs'].astype(str)
df['Installs'] = df['Installs'].str.replace("+", "", regex=False)
df['Installs'] = df['Installs'].str.replace("Free", "1", regex=False)
df['Installs'] = df['Installs'].str.replace(",", "", regex=False)
df['Installs'] = df['Installs'].astype(int)
df['Installs'].head(10)
```
